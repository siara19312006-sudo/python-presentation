```python
data = data.drop('Unnamed: 0', axis=1)
```

```python
data.head
```

```python
female_data = data[data['Gender'] == 'Female']
female_data.head()
```

```python
male_data = data[data['Gender'] == 'Male']
male_data.head()
```

```python
age_above_30 = data[data['Age'] > 30]
age_above_30.head()
```

```python
electronics_data = data[data['Product Category'] == 'Electronics']
electronics_data.head()
```

```python
high_sales = data[data['Total Amount'] > 500]
high_sales.head()
```

```python
female_beauty = data[
    (data['Gender'] == 'Female') &
    (data['Product Category'] == 'Beauty')
]

female_beauty.head()
```

```python
data['Date'] = pd.to_datetime(data['Date'])
```

```python
data.dtypes
```

```python
data['Year'] = data['Date'].dt.year
```

```python
data['Month'] = data['Date'].dt.month
```

```python
data['Day'] = data['Date'].dt.day
```

```python
data['Age Group'] = pd.cut(
    data['Age'],
    bins=[0, 18, 30, 45, 60, 100],
    labels=['Below 18', '18-30', '31-45', '46-60', 'Above 60']
)
```

```python
data.head
```

```python
gender_sales = data.groupby('Gender')['Total Amount'].sum()
gender_sales
```

```python
category_sales = data.groupby('Product Category')['Total Amount'].sum()
category_sales
```

```python
category_quantity = data.groupby('Product Category')['Quantity'].sum()
category_quantity
```

```python
age_group_sales = data.groupby('Age Group', observed=True)['Total Amount'].sum()
age_group_sales
```

```python
monthly_sales = data.groupby('Month')['Total Amount'].sum()
monthly_sales
```

```python
gender_avg_sales = data.groupby('Gender')['Total Amount'].mean()
gender_avg_sales
```

```python
category_avg_price = data.groupby('Product Category')['Price per Unit'].mean()
category_avg_price
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=data,
    x='Age',
    y='Total Amount',
    hue='Gender'
)

plt.title('Age vs Total Amount')
plt.xlabel('Age')
plt.ylabel('Total Amount')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x=gender_sales.index,
    y=gender_sales.values
)

plt.title('Sales by Gender')
plt.xlabel('Gender')
plt.ylabel('Total Sales')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x=category_sales.index,
    y=category_sales.values
)

plt.title('Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.lineplot(
    x=monthly_sales.index,
    y=monthly_sales.values,
    marker='o'
)

plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.show()
```

```python
plt.figure(figsize=(7,7))

plt.pie(
    category_sales.values,
    labels=category_sales.index,
    autopct='%1.1f%%'
)

plt.title('Sales Distribution by Product Category')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=data,
    x='Gender'
)

plt.title('Customer Count by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=data,
    x='Product Category'
)

plt.title('Number of Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=data,
    x='Age',
    bins=10,
    kde=True
)

plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=data,
    x='Product Category',
    y='Total Amount'
)

plt.title('Total Amount by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Amount')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.violinplot(
    data=data,
    x='Gender',
    y='Total Amount'
)

plt.title('Total Amount Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Total Amount')
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(
    data[['Age', 'Quantity', 'Price per Unit', 'Total Amount']].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(16, 10))

# 1. Sales by Product Category
plt.subplot(2, 2, 1)
sns.barplot(
    x=category_sales.index,
    y=category_sales.values
)
plt.title('Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')


# 2. Sales Distribution by Category
plt.subplot(2, 2, 2)
plt.pie(
    category_sales.values,
    labels=category_sales.index,
    autopct='%1.1f%%'
)
plt.title('Sales Distribution by Product Category')


# 3. Sales by Gender
plt.subplot(2, 2, 3)
sns.barplot(
    x=gender_sales.index,
    y=gender_sales.values
)
plt.title('Sales by Gender')
plt.xlabel('Gender')
plt.ylabel('Total Sales')


# 4. Monthly Sales Trend
plt.subplot(2, 2, 4)
sns.lineplot(
    x=monthly_sales.index,
    y=monthly_sales.values,
    marker='o'
)
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(16, 10))

# 1. Customer Count by Gender
plt.subplot(2, 2, 1)
sns.countplot(
    data=data,
    x='Gender'
)
plt.title('Customer Count by Gender')
plt.xlabel('Gender')
plt.ylabel('Customer Count')


# 2. Age Distribution
plt.subplot(2, 2, 2)
sns.histplot(
    data=data,
    x='Age',
    bins=10,
    kde=True
)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')


# 3. Total Amount by Product Category
plt.subplot(2, 2, 3)
sns.boxplot(
    data=data,
    x='Product Category',
    y='Total Amount'
)
plt.title('Total Amount by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Amount')


# 4. Total Amount by Gender
plt.subplot(2, 2, 4)
sns.violinplot(
    data=data,
    x='Gender',
    y='Total Amount'
)
plt.title('Total Amount Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Total Amount')

plt.tight_layout()
plt.show()
```

```python
# Monthly Quantity GroupBy
monthly_quantity = data.groupby('Month')['Quantity'].sum()

plt.figure(figsize=(18, 12))

# 1. Age vs Total Amount
plt.subplot(2, 3, 1)
sns.scatterplot(
    data=data,
    x='Age',
    y='Total Amount',
    hue='Gender'
)
plt.title('Age vs Total Amount')
plt.xlabel('Age')
plt.ylabel('Total Amount')


# 2. Correlation Heatmap
plt.subplot(2, 3, 2)
correlation = data[
    ['Age', 'Quantity', 'Price per Unit', 'Total Amount']
].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)
plt.title('Correlation Heatmap')


# 3. Quantity by Product Category
plt.subplot(2, 3, 3)
sns.barplot(
    x=category_quantity.index,
    y=category_quantity.values
)
plt.title('Quantity by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Quantity')


# 4. Average Price by Product Category
plt.subplot(2, 3, 4)
sns.barplot(
    x=category_avg_price.index,
    y=category_avg_price.values
)
plt.title('Average Price by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Average Price')


# 5. Sales by Age Group
plt.subplot(2, 3, 5)
sns.barplot(
    x=age_group_sales.index,
    y=age_group_sales.values
)
plt.title('Sales by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Total Sales')
plt.xticks(rotation=20)


# 6. Monthly Quantity Trend
plt.subplot(2, 3, 6)
sns.lineplot(
    x=monthly_quantity.index,
    y=monthly_quantity.values,
    marker='o'
)
plt.title('Monthly Quantity Trend')
plt.xlabel('Month')
plt.ylabel('Total Quantity')

plt.tight_layout()
plt.show()
```
